```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import numpy as np

# Generate data
X, y, coef = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=10,
                         coef=True,
                         random_state=0,
                         bias=100.0)

# Plot the data
plt.figure(figsize=(10, 6))
plt.scatter(X, y, alpha=0.6, label='Data points')

# Fit a LinearRegression model
reg = LinearRegression()
reg.fit(X, y)

# Get the equation of the fitted line
print(f"Equation: y = {reg.coef_[0]:.2f} * x + {reg.intercept_:.2f}")

# Add the fitted line to the plot
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_line = reg.predict(x_line)
plt.plot(x_line, y_line, color='red', label='Fitted line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Predict on X
y_pred = reg.predict(X)

# Create a function to compute Mean Squared Error
def compute_mse(y_true, y_pred):
    mse = np.mean((y_true - y_pred) ** 2)
    return mse

# Compute the MSE on the dataset
mse = compute_mse(y, y_pred)
print(f"MSE with noise=10: {mse:.2f}")

# Change noise parameter to 50 and repeat
X2, y2, coef2 = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=50,
                         coef=True,
                         random_state=0,
                         bias=100.0)

# Fit a LinearRegression model on new data
reg2 = LinearRegression()
reg2.fit(X2, y2)

print(f"Equation with noise=50: y = {reg2.coef_[0]:.2f} * x + {reg2.intercept_:.2f}")

# Predict on X2
y_pred2 = reg2.predict(X2)

# Compute the MSE on the new dataset
mse2 = compute_mse(y2, y_pred2)
print(f"MSE with noise=50: {mse2:.2f}")
```
