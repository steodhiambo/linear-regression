# Exercise 1: Scikit-learn estimator

The goal of this exercise is to learn to fit a Scikit-learn estimator and use it to predict.

X, y = [[1],[2.1],[3]], [[1],[2],[3]]

- Fit a LinearRegression from Scikit-learn with X the features and y the target and predict for x_pred = [[4]]
- Print the coefficients (coefs_) and the intercept (intercept_), the score (score) of the regression of X and y.

```python
# Exercise 1: Scikit-learn estimator

from sklearn.linear_model import LinearRegression
import numpy as np

# Define the data as specified in the exercise
X = [[1],[2.1],[3]]
y = [[1],[2],[3]]

# Create and fit the LinearRegression model
reg = LinearRegression()
reg.fit(X, y)

# Predict for x_pred = [[4]]
x_pred = [[4]]
y_pred = reg.predict(x_pred)

# Print the coefficients and intercept
print("Coefficients:", reg.coef_)
print("Intercept:", reg.intercept_)
print("Score:", reg.score(X, y))
print("Prediction for x_pred = [[4]]:", y_pred)
```
