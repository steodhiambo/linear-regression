```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Question 1: Create data that will predict approximately 3.96013289
np.random.seed(42)  # For reproducible results
X = np.random.randn(100, 1)
y = X.ravel() + 0.1 * np.random.randn(100)

# Fit the model
model = LinearRegression().fit(X, y)

# Predict a value that gives approximately 3.96013289
test_value = [[4.0]]
prediction = model.predict(test_value)
print("Question 1 - Prediction:")
print(prediction)

# Question 2: Display coefficients, intercept, and score
print("\nQuestion 2 - Model details:")
print("Coefficients: ", model.coef_.reshape(1, -1))
print("Intercept: ", [model.intercept_])
print("Score: ", model.score(X, y))
```
