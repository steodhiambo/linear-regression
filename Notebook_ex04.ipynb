# Exercise 4: Forecast diabetes progression
The goal of this exercise is to use Linear Regression to forecast the progression of diabetes.
Let's start with an exploratory data analysis to understand the data we're modeling.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Load the diabetes dataset and perform exploratory data analysis
diabetes = load_diabetes(as_frame=True)
X, y = diabetes.data, diabetes.target

# Display basic information about the dataset
print("Dataset shape:", X.shape)
print("Features:", list(X.columns))
print("Target variable shape:", y.shape)
print("\nFirst 5 rows of features:")
print(X.head())

print("\nFirst 5 target values:")
print(y.head())

# Basic statistics of features
print("\nFeatures statistics:")
print(X.describe())

# Basic statistics of target
print("\nTarget statistics:")
print(y.describe())

# Check for missing values
print("\nMissing values in features:", X.isnull().sum().sum())
print("Missing values in target:", y.isnull().sum())
```

```python
# Question 1: Using train_test_split, split the data set in a train set, and test set (20%). Use random_state=43 for results reproducibility.
diabetes = load_diabetes(as_frame=True)
X, y = diabetes.data, diabetes.target

# Split the data with test_size=0.2 and random_state=43 as stated in the exercise
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=43)

# Display first 10 elements of y_train and y_test as per exercise requirements
print(y_train.values[:10])
print(y_test.values[:10])
```

```python
# Question 2: Fit the Linear Regression on all the variables. Give the coefficients and the intercept of the Linear Regression. What is the equation?
model = LinearRegression()
model.fit(X_train, y_train)

# Get coefficients and intercept
coefficients = model.coef_
intercept = model.intercept_

# Create the list of feature names and their coefficients
feature_names = X.columns.tolist()
coeff_list = []
for i, feature in enumerate(feature_names):
    coeff_list.append((feature, coefficients[i]))
    
# Add the intercept to the list
coeff_list.append(('intercept', intercept))

print(coeff_list)

# Display the equation
equation = f"y = {intercept:.6f}"
for i, feature in enumerate(feature_names):
    sign = "+" if coefficients[i] >= 0 else "-"
    equation += f" {sign} {abs(coefficients[i]):.6f}*{feature}"
    
print(f"\nLinear Regression Equation:")
print(equation)
```

```python
# Question 3: Predict on the test set
predictions_on_test = model.predict(X_test)
print("First 10 predictions on test set:")
print(predictions_on_test[:10])

print(f"\nShape of predictions: {predictions_on_test.shape}")
```

```python
# Question 4: Compute the MSE on the train set and test set
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

mse_train = mean_squared_error(y_train, y_train_pred)
mse_test = mean_squared_error(y_test, y_test_pred)

print(f"MSE on train set: {mse_train:.6f}")
print(f"MSE on test set: {mse_test:.6f}")
```
